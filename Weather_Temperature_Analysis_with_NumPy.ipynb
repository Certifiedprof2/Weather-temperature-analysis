```python
## The manager wants the combine list of sales from both staffs
## Task 1: Try It With Python Lists

sales_timothy = [10, 15, 20]
sales_brenda = [5, 10, 15]

sales_timothy + sales_brenda

## Comments:
# 1. No, Python did not produce [15, 25, 35].
# 2. Python joined (concatenated) the two lists together.
# 3. This happened because + with Python lists means joining the lists, not adding corresponding elements.

```

```python
## Task 2: Convert Them to NumPy Arrays
sales_timothy_array = np.array(sales_timothy)
sales_brenda_array = np.array(sales_brenda)

sales_timothy_array + sales_brenda_array
```
